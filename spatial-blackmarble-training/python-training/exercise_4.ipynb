# Bonus: Make a Pretty Map using Nightlights

```python
import geopandas as gpd
from blackmarble import BlackMarble, Product, raster, extract
import rasterio
from dotenv import dotenv_values
from shapely.geometry import box
from pathlib import Path
import pandas as pd
```

```python
def find_repo_root(start=None):
    start = Path.cwd() if start is None else Path(start).resolve()
    for path in [start, *start.parents]:
        if (path / "data").exists() and (path / "spatial-blackmarble-training").exists():
            return path
    return start

REPO_ROOT = find_repo_root()
DATA_DIR = REPO_ROOT / "data"
RAW_DIR = DATA_DIR / "ntl_blackmarble" / "nigeria" / "raw"
RAW_DIR.mkdir(parents=True, exist_ok=True)

secrets_path = REPO_ROOT / ".config" / "ntl-training" / "secrets.env"
secrets = dotenv_values(secrets_path)
blackmarble_token = secrets.get("BLACKMARBLE_TOKEN", "").strip()
```

```python
bm = BlackMarble(
    token=blackmarble_token,
    output_directory=None,
    output_skip_if_exists=True,
)
```

```python
india = gpd.read_file('../../data/hdx/geoBoundaries-IND-ADM1_simplified.geojson')
india.explore()
```

```python
diwali_india = bm.raster(
    gdf = india,
    product_id = Product.VNP46A2,
    date_range = pd.date_range("2025-10-18", "2025-10-22", freq="D"),
)
```

```python
diwali_india = diwali_india.rio.write_crs("EPSG:4326")

for t in diwali_india.time.values:
    date_str = str(t)[:10]  # e.g. "2022-01"
    diwali_india["DNB_BRDF-Corrected_NTL"].sel(time=t).rio.to_raster(f"../../data/ntl_blackmarble/india/ntl_daily_{date_str}.tif")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import matplotlib.gridspec as gridspec
from scipy.ndimage import gaussian_filter

ntl_var = "DNB_BRDF-Corrected_NTL"
ntl = diwali_india[ntl_var]

# Custom city-lights colormap: black → ember → amber → gold → white
city_lights_colors = [
    "#000000", "#0a0500", "#1a0a00", "#3d1500",
    "#7a2e00", "#c45a00", "#e88a00", "#f5c842",
    "#fde68a", "#fff8dc", "#ffffff",
]
cmap = mcolors.LinearSegmentedColormap.from_list("city_lights", city_lights_colors, N=512)

all_vals = np.log1p(ntl.values)
vmin = 0
vmax = np.nanpercentile(all_vals[np.isfinite(all_vals)], 99.5)

times = ntl.time.values
n = len(times)  # 5

fig = plt.figure(figsize=(6 * n, 7), facecolor="#000000")
fig.suptitle(
    "Nighttime Lights — India  •  Oct 18–22 2025",
    color="#f5c842", fontsize=16, fontweight="bold",
    y=1.01,
)

gs = gridspec.GridSpec(
    1, n, figure=fig,
    hspace=0, wspace=0.03,
    left=0.02, right=0.98,
    top=0.95, bottom=0.08,
)

for i, t in enumerate(times):
    ax = fig.add_subplot(gs[0, i])
    date = pd.Timestamp(t)
    day_label = date.strftime("%b %d")
    is_diwali = date.day == 20  # Oct 20 2025 is Diwali

    arr = np.log1p(ntl.sel(time=t).values)

    # Base layer — sharp raster
    ax.imshow(
        arr,
        cmap=cmap, vmin=vmin, vmax=vmax,
        origin="upper",
        interpolation="bilinear",
        aspect="auto",
    )

    # Bloom layer — blurred glow on top
    glow = gaussian_filter(np.nan_to_num(arr, nan=0.0), sigma=3)
    ax.imshow(
        glow,
        cmap=cmap, vmin=vmin, vmax=vmax,
        origin="upper",
        interpolation="bilinear",
        aspect="auto",
        alpha=0.45,
    )

    ax.set_axis_off()
    ax.set_facecolor("#000000")

    # Date label
    label_color = "#ffffff" if is_diwali else "#f5c842"
    ax.text(
        0.5, 0.04, day_label,
        transform=ax.transAxes,
        color=label_color, fontsize=13, fontweight="bold",
        ha="center", va="bottom",
        bbox=dict(boxstyle="round,pad=0.3", facecolor="#000000", alpha=0.6, edgecolor="none"),
    )

# Shared colorbar
cbar_ax = fig.add_axes([0.15, 0.03, 0.70, 0.018])
sm = plt.cm.ScalarMappable(cmap=cmap, norm=plt.Normalize(vmin=vmin, vmax=vmax))
sm.set_array([])
cbar = fig.colorbar(sm, cax=cbar_ax, orientation="horizontal")
cbar.set_label("log(1 + radiance)  [nW·cm⁻²·sr⁻¹]", color="#888888", fontsize=9, labelpad=5)
plt.setp(cbar.ax.xaxis.get_ticklabels(), color="#888888", fontsize=8)
cbar.outline.set_edgecolor("#333333")

plt.savefig("diwali_ntl.png", dpi=200, bbox_inches="tight", facecolor="#000000")
plt.show()
```
